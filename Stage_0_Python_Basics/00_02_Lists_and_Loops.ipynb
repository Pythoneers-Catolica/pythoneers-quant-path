## Topic 1: Creating a List

When we want to group related values together, we turn to a Python **list**. Think of a list like a labeled portfolio folder or a grocery checklist. It keeps our items organized in a specific order so we can easily find them later.

Each individual value inside a list is called an **element**. To build a list, we wrap our elements in square brackets `[` and `]`, separating entries with a comma:
- **Square Brackets `[` and `]`**: Define the boundaries of our list.
- **Commas `,`**: Separate adjacent elements from each other.

Let's build our first list of stock tickers and inspect what it looks like. Let's see:

````python
# We bundle three tech tickers into a single list named tickers
tickers = ["AAPL", "MSFT", "GOOGL"]

# Let's inspect our entire list on screen
print(tickers)
````

> **Key Takeaway**: A list holds multiple ordered elements inside square brackets, separated cleanly by commas.

---

## Topic 2: Accessing Items by Index

Once we store several tickers inside a list, we frequently need to pull out just one of them. To do that, we use an address called an **index**, which indicates the exact position of an element.

Here is where programming can feel a bit counter-intuitive at first. In Python, counting always starts at **0** rather than 1. This system is known as **zero-based indexing**. If this feels strange right now, do not worry. Every programmer has to retrain their brain on this concept!

Think of the index as the number of steps away from the starting line:
- **Index 0**: We take zero steps from the start, landing directly on our first item (`"AAPL"`).
- **Index 1**: We take one step forward to reach our second item (`"MSFT"`).
- **Index 2**: We take two steps forward to reach our third item (`"GOOGL"`).

To retrieve an item, we write the list name followed by square brackets holding the index position, such as `tickers[0]`.

Let's retrieve our first ticker. Let's see:

````python
# We access the first element at index 0
print(tickers[0])
````

````python
# Now we retrieve the third element by requesting index 2
print(tickers[2])
````

### Negative Indexing: Counting Backward from the End

Imagine we track a large portfolio containing hundreds of stocks and we need to check the very last one added. Calculating the exact length of the list just to find that final position would be tedious.

Python gives us a handy shortcut called **negative indexing**. When we supply a negative number, Python counts backward from the end of the line:
- **Index -1**: Reaches the final element in the collection.
- **Index -2**: Reaches the second-to-last element.

This shortcut is especially popular in finance when we want the most recent closing price without doing extra math.

Let's grab the final ticker from our list. Let's check:

````python
# Using index -1 lets us pick the final ticker directly
print(tickers[-1])
````

> **Key Takeaway**: Python uses zero-based indexing for forward counting, while negative indices let us count backward from the end.

## Topic 3: Modifying a List

Portfolios are rarely frozen in place. As market conditions evolve, we welcome new stocks to our watchlists and replace assets that no longer fit our strategy.

Fortunately, Python lists are **mutable**, which simply means we can change them after they are created. We can append fresh entries to the tail of our list or overwrite an existing element whenever we want.

### Adding an Item with .append()

When we want to add an element to the end of an existing list, we use the `.append()` method. Think of this like writing an extra stock symbol onto the bottom of our trading notepad.

Let's append `"NVDA"` to our watchlist and verify how the list changes. Let's see:

````python
# We append NVDA to the end of our existing tickers list
tickers.append("NVDA")

# Let's inspect the updated list to confirm our new addition
print(tickers)
````

### Updating an Existing Value

What happens when we want to swap out an asset rather than append to the end? We can overwrite any position by targeting its index and using the assignment operator `=`.

Think of this like erasing a ticker from a specific line on our clipboard and writing a new one in its place. The rest of our list stays completely intact.

Let's swap out the second item at index 1 (`"MSFT"`) and replace it with `"AMZN"`. Let's check:

````python
# We assign AMZN to index position 1, replacing MSFT
tickers[1] = "AMZN"

# Let's inspect the list to verify the update
print(tickers)
````

> **Key Takeaway**: Lists are mutable, allowing us to append new elements to the end or modify specific positions by index.

## Topic 4: The for Loop

Imagine we track one hundred stock symbols and need to print each one. Typing one hundred separate `print()` statements would be slow and prone to mistakes. Whenever we catch ourselves repeating the same line of code, there is almost always a better way.

This is why we use a **for loop**. A loop acts like an automated conveyor belt. Python picks up each item from our collection and runs our instructions on it before moving to the next item until the list is complete.

Each complete pass through our instructions is called an **iteration**.

### Anatomy of a for Loop

Here is the standard structure we use:

```python
for ticker in tickers:
    print(ticker)
```

Let's look at how these pieces work together:
- **`for` and `in`**: These keywords tell Python that we want to visit every element in our collection from start to finish.
- **`ticker`**: This is our loop variable. On every pass, Python automatically assigns the current element to this temporary name. We can choose any name we like, though a descriptive word keeps our code easy to read.
- **`tickers`**: The list holding our data.
- **The Colon `:`**: This punctuation mark signals that our loop header is complete and that the repeating instructions follow immediately below.
- **Indentation (4 Spaces)**: Python relies on indented spaces to know which lines belong inside the loop body.

Missing a colon or forgetting to indent by four spaces happens to everyone when first starting out. With a little practice, your hands will build this habit naturally.

Let's watch our loop visit each stock in our list. Let's see:

````python
# We step through our list and display each ticker one by one
for ticker in tickers:
  print(ticker)
````

### Why Indentation Matters

In many programming languages, developers use curly braces to mark the boundaries of a loop. In Python, clean visual spacing serves as our grammar:
- Any line indented by **4 spaces** runs repeatedly on every single iteration of the loop.
- Any unindented line tells Python that the loop has finished, returning to normal top-to-bottom execution.

Think of indentation like sub-tasks indented under a main heading on a checklist. If an action is indented, it belongs to the loop cycle.

Let's test this behavior with an analysis message that prints during the loop, followed by a final completion message that runs once the loop wraps up. Let's check:

````python
for ticker in tickers:
  # Indented lines execute repeatedly on every single pass of the loop
  print(f"Analyzing stock: {ticker}")

# Unindented lines run only once after the entire loop has completed
print("All stocks have been analyzed!")
````

> **Key Takeaway**: A `for` loop automates repetitive tasks. Indenting by 4 spaces defines which actions repeat, while unindented code runs once after the loop finishes.

## Topic 5: Looping with range() and len()

In our previous loop, Python handed us the value of each stock directly. But in quantitative finance, we often want to know the asset's numerical position or ranking as well. For instance, we might want to display rank numbers alongside our portfolio holdings.

To track both the position and the value, we pair two built-in Python helpers:
- **`len(tickers)`**: Short for length. It counts the total number of elements inside our list.
- **`range(n)`**: Generates an ordered sequence of counting numbers starting at 0 and stopping just before `n`.

Seeing nested parentheses like `range(len(tickers))` can look intimidating at first glance. If you ever feel stuck, read them from the inside out: first `len(tickers)` counts four items, and then `range(4)` generates the position numbers from `0` up to `3`.

We commonly use the variable name `i` (short for *index*) to walk through these position numbers one by one.

Let's see how this works in practice:

````python
# We loop through index numbers from 0 up to len(tickers) - 1
for i in range(len(tickers)):
  # i represents the current index, while tickers[i] retrieves the element at that spot
  print(i, tickers[i])
````

### Polishing Output with f-strings

Raw numbers and symbols are functional, but in real trading systems we want our summaries to look crisp and professional. We can blend our loop index with the f-strings we learned in Workshop 0.1 to build a neat ranking display.

Let's see how cleanly we can present our portfolio positions:

````python
# We format each position index alongside its corresponding stock ticker
for i in range(len(tickers)):
  print(f"Position {i}: {tickers[i]}")
````

> **Key Takeaway**: Pairing `range()` with `len()` allows us to loop through both the numerical index and the element value at the same time.
