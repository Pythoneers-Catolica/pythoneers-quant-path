### Teaching Programs to Make Decisions

In live quantitative trading, our software cannot simply run a rigid sequence of instructions from top to bottom. Instead, algorithms must constantly evaluate market conditions and choose how to respond. In computer programming, this ability to make choices is called **conditional logic**. We ask Python a question about our data, and depending on whether the answer is true or false, our code follows a specific branch of instructions.

In this workshop, we will explore how to test comparisons, guide program flow with `if` statements, and combine multiple market rules together.

> **Key Takeaway**: Conditional logic lets our programs make dynamic choices based on whether market rules evaluate to true or false.

## Topic 1: Booleans and Comparisons

Before a program can choose between different actions, it must evaluate whether a statement holds true. Python represents the outcome of these questions using a **Boolean** data type, named in honor of mathematician George Boole.

A Boolean holds one of only two possible truth states: `True` or `False`. Notice that both values always begin with a capital letter.

We ask questions about our data using **comparison operators**:
- `>` and `<`: Greater than, and less than.
- `>=` and `<=`: Greater than or equal to, and less than or equal to.
- `==`: Equal to (notice the double equal sign).
- `!=`: Not equal to (the exclamation mark represents negation).

Mixing up `=` and `==` is easily the most common stumbling block for newcomers. Remember that a single `=` assigns a value into a variable box, while a double `==` asks a question comparing two values.

Let's test these comparisons against a stock price. Let's see:

````python
price = 150.25

# Question 1: Is price greater than 100?
print(price > 100)

# Question 2: Is price exactly equal to 150.25? (notice ==)
print(price == 150.25)

# Question 3: Is price NOT equal to 150.0?
print(price != 150.0)
````

Now let's check what comparisons look like when the condition fails to hold true. Let's check:

````python
# Is price less than 50?
print(price < 50)

# Is price equal to 200.0?
print(price == 200.0)
````

> **Key Takeaway**: Comparison operators test relationships between values, returning either `True` or `False`.

## Topic 2: The if Statement

An **if statement** functions like a checkpoint or tollgate in our program. Python tests our condition, and if the result is `True`, it steps inside to execute the indented code block.

### Anatomy of an if Statement

Here is the fundamental structure:

```python
if price > 100:
    print("This is an expensive stock")
```

This statement relies on three essential elements:
- The `if` keyword opens the decision branch.
- A trailing colon `:` confirms that our condition has been fully specified.
- Indented instructions (pushed right by four spaces) tell Python exactly which statements belong to this decision block.

Let's see our first `if` statement evaluate an active stock price. Let's see:

````python
price = 150.25

# Because 150.25 > 100 evaluates to True, this indented block executes
if price > 100:
  print("This is an expensive stock")
````

### What Happens When the Condition is False?

When a condition evaluates to `False`, Python quietly bypasses the indented block and continues running the unindented code that follows.

Let's test this behavior with a lower stock price. Let's check:

````python
cheap_price = 25.0

# Because 25.0 > 100 is False, Python skips this indented print statement
if cheap_price > 100:
  print("This will not print!")

# This unindented line always runs regardless of the condition above
print("Condition checked.")
````

> **Key Takeaway**: An `if` statement executes its indented body only when its test condition evaluates to `True`.

## Topic 3: else and elif

Real trading strategies require backup plans. If a condition is not met, we usually want an alternative response rather than doing nothing at all. This is where `else` and `elif` come in.

Think of this structure like a series of railway switches:
- **`if`**: Tests our primary condition first.
- **`elif`**: Short for *else if*. It tests an alternative condition only when earlier checks evaluated to `False`.
- **`else`**: Serves as our catch-all fallback, executing only when none of the preceding conditions were met.

Let's see this decision chain categorize a stock into price tiers. Let's see:

````python
price = 150.25

if price > 200:
  print("Very expensive")
elif price > 100:
  print("Moderately priced")
else:
  print("Cheap stock")
````

### Understanding How Python Evaluates Branches

Python evaluates conditional branches from top to bottom. It tests the initial condition `150.25 > 200`, which turns out to be `False`. It then steps to the `elif` branch and evaluates `150.25 > 100`, which is `True`.

Once Python executes a successful branch, it leaves the entire structure immediately and skips any remaining fallback blocks.

Let's observe what happens when we feed in a stock price of `30.0`. Let's check:

````python
price = 30.0

if price > 200:
  print("Very expensive")
elif price > 100:
  print("Moderately priced")
else:
  print("Cheap stock")
````

> **Key Takeaway**: Combining `if` with alternative branches creates orderly decision trees where only the first matching branch executes.

## Topic 4: Combining Conditions with and and or

Quantitative trading rules often demand that multiple signals align before taking action. For instance, we might require that an asset is attractively priced while simultaneously trading with robust volume.

We can join separate comparisons using the logical operators `and` and `or`.

### The and Operator

With `and`, every individual condition must evaluate to `True` for the combined statement to succeed. If any part fails, the entire expression evaluates to `False`.

Let's test whether our stock satisfies both price and volume criteria simultaneously. Let's see:

````python
price = 150.25
volume = 2000000

# Both price > 100 and volume > 1000000 evaluate to True
if price > 100 and volume > 1000000:
  print("High price and high volume")
````

### The or Operator

With `or`, the expression evaluates to `True` whenever at least one condition is satisfied. It only results in `False` when every single condition fails.

Let's see an alert that triggers if a stock is cheap or if its trading volume is unusually high. Let's check:

````python
price = 45.0
volume = 500000

# Even though volume is low, price < 50 evaluates to True
if price < 50 or volume > 1000000:
  print("Alert: Interesting opportunity!")
````

> **Key Takeaway**: Use `and` when every condition must hold true, and use `or` when satisfying any single condition is enough.
