# Workshop 0.5: Files and Paths

Welcome to Workshop 0.5! Up to this point, we have stored stock prices and watchlist collections inside Python variables. While that works smoothly during an active session, variables exist only in computer memory.

### Preserving Market Data on Disk

Think of computer memory (RAM) like a dry-erase whiteboard. It provides fast workspace while your program is running, but as soon as you shut down Python or close your notebook, the whiteboard gets wiped completely clean.

In quantitative research, we cannot afford to lose our historical prices or computed signals every time a program stops. We need permanent storage on our disk, much like storing physical ledgers inside an archival filing cabinet.

Modern Python makes interacting with files intuitive through the built-in `pathlib` module. In this workshop, we will explore how to import tools, set up project folders, and write data directly to disk.

> **Key Takeaway**: Memory is volatile and clears when programs stop, whereas writing to disk preserves our datasets permanently.

## Topic 1: Importing a Module

Python includes a vast ecosystem of capabilities, but loading every tool at once would slow down our computer. Instead, Python keeps specialized toolkits packaged into **modules**, which we bring in using the `import` statement.

One of the most practical standard modules is `pathlib`. Inside it lives the `Path` class, which gives us an object-oriented way to represent file and directory locations across any operating system.

Let's import the `Path` tool and verify that it loads cleanly. Let's see:

````python
# Import the Path tool from the built-in pathlib module
from pathlib import Path

print("Path tool loaded successfully!")
````

> **Key Takeaway**: We use `from module import tool` to bring specialized capabilities into our script only when needed.

---

## Topic 2: Creating a Folder

In professional quant workflows, tidy organization is essential. We avoid dumping raw data, cache files, and reports into a single disorganized pile. Instead, we create dedicated directories for each stage of our research pipeline.

With `Path`, making a new folder involves defining its location and calling `.mkdir()` (short for *make directory*). We commonly pass `exist_ok=True` so that Python quietly proceeds if the folder already exists rather than raising an error.

Let's build a dedicated `data/` directory. Let's see:

````python
# First, we define the folder location:
data_folder = Path("data/")

# Next, we create the folder on disk if it does not already exist:
data_folder.mkdir(exist_ok=True)

print(f"Folder created: {data_folder}")
````

> **Key Takeaway**: Calling `.mkdir(exist_ok=True)` safely creates directory folders on disk without throwing errors if they already exist.

---

## Topic 3: Writing to a File

Now that our folder is ready, we can place text files inside it. In `pathlib`, the forward slash operator `/` joins folder paths and filenames together seamlessly.

Once we have our path, we call `.write_text()` to record our information onto disk. When saving multiple entries, we separate them with the newline escape character `\n`, which simulates pressing the return key.

Let's record three ticker symbols into a text file. Let's see:

````python
# We join the folder and file name cleanly using the slash operator:
file_path = data_folder / "tickers.txt"

# Then we write our text directly to disk using utf-8 encoding:
file_path.write_text("AAPL\nMSFT\nGOOGL")

print(f"Wrote tickers to: {file_path}")
````

> **Key Takeaway**: We join folder paths and filenames using `/`, and save raw text to disk using `.write_text()`.

---

## Topic 4: Reading from a File

Retrieving saved data back into memory is just as direct. Calling `.read_text()` on any existing path reads the contents from disk and returns them as a standard Python string.

Let's pull our saved tickers back from disk to verify what was stored. Let's check:

````python
# Read the contents of the file from disk:
content = file_path.read_text()

# Print the loaded content:
print(content)
````

> **Key Takeaway**: `.read_text()` loads the full content of a file directly off disk into a Python string.

---

## Topic 5: Writing a List to a File

In trading workflows, assets frequently arrive as lists of strings, like `tickers = ["AAPL", "MSFT", "NVDA"]`. However, `.write_text()` expects a single text string rather than a list container.

To bridge this gap, we use the string method `"\n".join(tickers)`. This command glues each element of our list together, placing a newline character between adjacent items.

Let's convert our list into formatted text and write it out in one clean operation. Let's check:

````python
tickers = ["AAPL", "MSFT", "NVDA"]

# Join list items with a newline between each:
content = "\n".join(tickers)

# Write the joined string to the file in one go:
file_path.write_text(content)

# Read it back and print to confirm:
print("File updated! Current contents:")
print(file_path.read_text())
````

> **Key Takeaway**: Use `"\n".join(items)` to turn list entries into a clean multi-line string before writing to disk.
