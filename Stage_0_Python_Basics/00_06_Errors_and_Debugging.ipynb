### Reframing How We Look at Errors

Every programmer, from someone writing their very first script to a seasoned quant developer at a top fund, encounters errors daily. Seeing a red error block can feel discouraging when you first start, but errors are actually your most valuable learning tool.

An error is not Python saying you failed. It is simply Python being an honest assistant: *"I tried to execute your code, but I ran into a puzzle on this specific line."* Think of error messages like diagnostic check-engine lights or GPS alerts rerouting you when a road is closed.

In this workshop, we will learn how to read error reports without fear, handle unexpected conditions gracefully with `try` and `except`, and resolve common beginner slip-ups.

> **Key Takeaway**: Errors are diagnostic clues rather than personal failures. Learning to interpret them is how we become capable quantitative developers.

## Topic 1: Reading a Traceback

When Python encounters an instruction it cannot complete, it pauses execution and prints a diagnostic summary known as a **traceback**.

A traceback can look overwhelming because it shows the entire call sequence, but you only need to look at two key spots:
- Start at the very bottom line. The final line identifies the **error type** alongside a short description of the problem.
- Look directly above that line for the line number indicator. Python points an arrow directly at the line where execution stopped.

Let's intentionally divide a number by zero to observe Python's diagnostic report. Let's see:

````python
# Intentionally dividing by zero:
result = 10 / 0
````

### Deconstructing the ZeroDivisionError

Notice how cleanly Python explains the problem. The final line tells us that division by zero is mathematically undefined, while the arrow points directly to line 2.

Now let's examine another frequent situation: requesting a key that does not exist in our dictionary. Let's see:

````python
prices = {"AAPL": 150}
# Looking up 'MSFT' which is not in the dictionary:
print(prices["MSFT"])
````

### Deconstructing the KeyError

Here, Python reports a `KeyError: 'MSFT'`. This message tells us that Python searched through `prices` but found no entry under that label. Whenever you see a `KeyError`, verify your spelling or confirm that the symbol was properly inserted.

> **Key Takeaway**: Always read tracebacks from the bottom up to identify the error type and locate the exact line where execution stalled.

## Topic 2: try and except: Catching Errors

In live quantitative trading, we never want an entire strategy to crash simply because a single asset was missing from an incoming market feed. We need a way to catch issues smoothly and keep the rest of our pipeline running.

Python provides **`try` and `except`** blocks, which function like circuit breakers in an electrical panel:
- The code inside the **`try`** block runs first.
- If everything executes without issue, Python bypasses the **`except`** block entirely.
- If a targeted error occurs, Python catches it instantly and runs our fallback instructions instead of crashing.

Let's see how a `try` and `except` block shields our dictionary lookup. Let's check:

````python
try:
  prices = {"AAPL": 150}
  print(prices["MSFT"])
except KeyError:
  print("That ticker was not found!")
````

Notice what happened here: no red traceback appeared, and our program completed cleanly. It intercepted the missing key and displayed our friendly fallback message.

> **Key Takeaway**: Wrapping risky operations inside `try` and `except` prevents unexpected data glitches from crashing your pipeline.

## Topic 3: Catching Different Error Types

Different programming mistakes trigger different error classes. For instance, a **ValueError** occurs when a function receives an input of the expected type but carrying an unsuitable value, such as trying to convert the letters `"abc"` into an integer.

Let's see how we can handle a conversion error gracefully. Let's see:

````python
try:
  value = int("abc") # This will cause ValueError
except ValueError:
  print("Could not convert to integer.")
````

### Using a General Exception as a Safety Net

When you are unsure of the specific error that might emerge, Python lets you catch standard exceptions using `Exception as e`. The keyword `as e` stores the underlying error details inside variable `e` so you can inspect them.

Let's observe our general safety net in action. Let's check:

````python
try:
  result = 10 / 0
except Exception as e:
  print("Something went wrong:", e)
````

> **Key Takeaway**: We can catch specific errors like `ValueError` or use `Exception as e` to log diagnostic messages without interrupting program flow.

## Topic 4: Why Errors Happen (Common Beginner Mistakes)

Let's explore two of the most frequent errors encountered by developers so you can recognize and resolve them quickly.

### NameError (Misspelled Variable Name)

A `NameError` occurs when you reference a variable name that Python does not recognize. This issue almost always stems from a simple typo or forgetting to run the cell where the variable was defined.

Let's see a `NameError` caused by an extra letter in our variable name. Let's see:

````python
price = 150.0
# Typo: 'pricee' instead of 'price'
print(pricee)
````

### Resolving a NameError

To resolve a `NameError`, double-check the spelling of your variable against where it was defined. In Jupyter notebooks, also make sure you ran the cell that created the variable earlier in the notebook.

### IndexError (Index Beyond Collection Size)

An `IndexError` happens when you ask Python for a position that falls outside the boundaries of your list.

Let's see an `IndexError` when requesting index 5 on a list of two elements. Let's see:

````python
lst = [1, 2]
# lst only has indexes 0 and 1; index 5 does not exist:
print(lst[5])
````

### Resolving an IndexError

Remember that Python uses zero-based indexing: a list with two items only offers indices 0 and 1. You can call `len(lst)` to check how many elements exist before requesting a specific position.

> **Key Takeaway**: `NameError` usually indicates a typo or unexecuted cell, while `IndexError` warns that your requested position exceeds the list boundaries.
