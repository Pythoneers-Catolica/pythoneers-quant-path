## Topic 1: Creating a Series (1-Dimensional Data)

The fundamental building block of pandas is the **Series**. A Series represents a one-dimensional labeled array, much like a single column in a financial ledger.

Every Series combines two essential parts:
- **Values**: The actual underlying observations, such as prices or volumes.
- **Index**: An explicit label or row identifier assigned to each value, defaulting to `0, 1, 2, ...`.

Let's create our first Series from a collection of prices. Let's see:

````python
# Create a Series from a list of numbers:
prices = pd.Series([150, 152, 149, 153])

# Print the Series:
print(prices)
````

### Inspecting Series Components

Notice how pandas formats the output: the row index appears on the left, while the recorded values sit on the right. At the bottom, `dtype: int64` tells us that pandas stored our values as 64-bit integers.

We can inspect individual components directly:
- `prices[0]`: Retrieves an item using its index position.
- `prices.values`: Extracts the raw NumPy array.
- `prices.index`: Returns the index range metadata.

Let's see these properties in action. Let's check:

````python
# We access the first element using zero-based index 0:
print("First item:", prices[0])

# We extract the underlying NumPy array of values:
print("Values:", prices.values)

# We inspect the index labels:
print("Index:", prices.index)
````

> **Key Takeaway**: A Series represents a single labeled column, coupling an array of values with an explicit index.

## Topic 2: Creating a DataFrame from a Dictionary

While a Series represents a single column, financial tables typically contain multiple columns side-by-side. When we align multiple Series along a shared index, we form a **DataFrame**.

A DataFrame is a two-dimensional grid with labeled rows and named columns, much like an Excel sheet or SQL table.

We can build a DataFrame easily from a Python dictionary. Each dictionary key becomes a column header, and the associated list forms the column observations.

Let's assemble a product inventory table. Let's see:

````python
# We bundle our data into a dictionary where keys become column names:
data = {
  "Product": ["Laptop", "Monitor", "Keyboard", "Mouse"],
  "Price": [1200, 300, 50, 25],
  "Quantity": [5, 10, 20, 15]
}

# We convert our dictionary into a two-dimensional DataFrame:
df = pd.DataFrame(data)

# Let's inspect the tabular output:
print(df)
````

Notice how cleanly pandas aligns our dictionary into a structured grid, automatically attaching a default numerical index on the left.

> **Key Takeaway**: A DataFrame organizes tabular data into rows and named columns, turning standard dictionaries into interactive tables.

## Topic 3: Exploring a DataFrame

In real-world quantitative systems, datasets can span millions of records. Printing an entire table at once is impractical, so pandas provides swift inspection methods:
- **`.head(n)`**: Displays the first `n` rows of our table, offering an immediate visual sanity check.
- **`.info()`**: Summarizes column names, non-null counts, and memory consumption.
- **`.describe()`**: Calculates descriptive statistics across all numerical columns, including averages and percentiles.
- **`.shape`**: Reports overall dimensions as a tuple `(rows, columns)`. Notice that `.shape` is an attribute rather than a method, so it does not use parentheses.

Let's inspect our table using these diagnostic tools. Let's see:

````python
# We preview the first two rows:
print("--- df.head(2) ---")
print(df.head(2))

# We inspect column types and non-null counts:
print("\n--- df.info() ---")
df.info()

# We compute quick summary statistics:
print("\n--- df.describe() ---")
print(df.describe())

# We check table dimensions (rows, columns):
print("\n--- df.shape ---")
print(df.shape)
````

> **Key Takeaway**: Use `.head()`, `.info()`, `.describe()`, and `.shape` to diagnose the structure and health of any DataFrame quickly.

## Topic 4: Accessing a Column

When extracting specific columns from a DataFrame, we choose between two main syntaxes:
- **Square Bracket Notation `df["Price"]`**: The standard and safest approach, returning the targeted column as a pandas Series.
- **Dot Notation `df.Price`**: A handy shorthand that behaves identically. Keep in mind that dot notation fails if a column name contains spaces or collides with built-in pandas methods.

If we want to select multiple columns at once, we use **double square brackets** `df[["Product", "Price"]]`. The outer brackets signal column extraction, while the inner brackets supply a list of column names, returning a subset DataFrame.

Let's examine single and multi-column selections. Let's see:

````python
# We select a single column using bracket notation:
print("--- Using Square Brackets df['Price'] ---")
print(df["Price"])

# We can also select the column as an attribute:
print("\n--- Using Dot Notation df.Price ---")
print(df.Price)

# We select multiple columns by passing a list of column names:
print("\n--- Multiple Columns df[['Product', 'Price']] ---")
print(df[["Product", "Price"]])
````

> **Key Takeaway**: Single brackets extract a Series, whereas double brackets extract a multi-column DataFrame.

---

## Topic 5: Adding a New Column with Vectorized Math

In Stage 0, calculating total inventory values across multiple items required writing a `for` loop to multiply each row individually. In pandas, we rarely need loops for column calculations.

Instead, pandas supports **vectorized operations**. When we write `df["Price"] * df["Quantity"]`, pandas carries out row-by-row multiplication across the entire column simultaneously in compiled C code.

Assigning the expression with `=` appends the computed result as a brand new column in our DataFrame.

Let's compute our inventory totals in a single vectorized statement. Let's check:

````python
# We compute a new 'Total' column using vectorized multiplication:
df["Total"] = df["Price"] * df["Quantity"]

# Let's check the updated table:
print(df)
````

> **Key Takeaway**: Vectorized arithmetic executes column operations simultaneously at C-speed, removing the need for manual row loops.
